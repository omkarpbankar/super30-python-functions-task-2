## Question 1: Sum of Variable Arguments using `*args`

### Objective:
Create a function using `*args` that accepts any number of values and returns their total.

### Concept:
`*args` allows a function to accept an arbitrary number of positional arguments. Inside the function, `args` behaves as a tuple containing all supplied arguments.

```python
def calculate_total(*args: float | int) -> float | int:
    """
    Accepts any number of positional numeric arguments and returns their sum.
    """
    total = 0
    for num in args:
        total += num
    return total

# Testing calculate_total
print(f"calculate_total(10, 20, 30)        -> {calculate_total(10, 20, 30)}")
print(f"calculate_total(5, 10, 15, 20, 25) -> {calculate_total(5, 10, 15, 20, 25)}")
print(f"calculate_total(1.5, 2.5, 3.0)     -> {calculate_total(1.5, 2.5, 3.0)}")
print(f"calculate_total()                  -> {calculate_total()}")
```

## Question 2: Find Largest Number using `*args`

### Objective:
Create a function using `*args` that returns the largest supplied number.

### Concept:
Iterate through the tuple unpacked by `*args`, keeping track of the maximum value found. Handle edge cases like empty arguments gracefully.

```python
def find_max(*args: float | int) -> float | int | None:
    """
    Accepts any number of positional numeric arguments and returns the largest one.
    """
    if not args:
        return None
    
    max_val = args[0]
    for num in args[1:]:
        if num > max_val:
            max_val = num
    return max_val

# Testing find_max
print(f"find_max(10, 45, 2, 99, 34) -> {find_max(10, 45, 2, 99, 34)}")
print(f"find_max(-15, -4, -80, -2)  -> {find_max(-15, -4, -80, -2)}")
print(f"find_max(42)                -> {find_max(42)}")
print(f"find_max()                  -> {find_max()}")
```

## Question 3: Dynamic User Information using `**kwargs`

### Objective:
Create `def create_profile(**kwargs)` that accepts dynamic user information and prints all provided attributes.

### Concept:
`**kwargs` captures arbitrary keyword arguments into a Python dictionary, allowing dynamic attribute passing at runtime.

```python
from typing import Any

def create_profile(**kwargs: Any) -> dict[str, Any]:
    """
    Accepts dynamic user information as keyword arguments and prints formatted profile.
    """
    print("\n" + "=" * 40)
    print("           USER PROFILE")
    print("=" * 40)
    if not kwargs:
        print("  No profile attributes provided.")
    else:
        for key, value in kwargs.items():
            formatted_key = key.replace("_", " ").title()
            print(f"  * {formatted_key:<16}: {value}")
    print("=" * 40)
    return kwargs

# Testing create_profile
profile = create_profile(
    name="Omkar Bankar",
    role="AI Engineer",
    skills=["Python", "Deep Learning", "FastAPI"],
    city="Pune",
    experience="3 years"
)
```

## Question 4: Higher-Order Functions (Passing Functions as Arguments)

### Objective:
Create a function `calculate(func, a, b)` that accepts another function as an argument.

**Example Idea:**
- `calculate(add, 10, 20)`
- `calculate(multiply, 10, 20)`

### Concept:
In Python, functions are *first-class citizens*. They can be assigned to variables, passed as parameters into other functions, and returned from functions.

```python
from typing import Callable

def add(a: float | int, b: float | int) -> float | int:
    """Returns sum of a and b."""
    return a + b

def subtract(a: float | int, b: float | int) -> float | int:
    """Returns difference between a and b."""
    return a - b

def multiply(a: float | int, b: float | int) -> float | int:
    """Returns product of a and b."""
    return a * b

def divide(a: float | int, b: float | int) -> float | str:
    """Returns quotient of a and b with zero division check."""
    if b == 0:
        return "Error: Division by zero is not allowed."
    return a / b

def calculate(func: Callable[[float | int, float | int], Any], a: float | int, b: float | int) -> Any:
    """
    Higher-order function that applies func to arguments a and b.
    """
    return func(a, b)

# Testing calculate with various functions
print(f"calculate(add, 10, 20)      -> {calculate(add, 10, 20)}")
print(f"calculate(subtract, 50, 15) -> {calculate(subtract, 50, 15)}")
print(f"calculate(multiply, 10, 20) -> {calculate(multiply, 10, 20)}")
print(f"calculate(divide, 100, 4)   -> {calculate(divide, 100, 4)}")
print(f"calculate(divide, 50, 0)    -> {calculate(divide, 50, 0)}")
```

## Question 8: Recursive Factorial Function

### Objective:
Write a recursive function to calculate factorial ($n!$).

### Mathematical Definition:
$$
n! = \begin{cases}
1, & \text{if } n = 0 \text{ or } n = 1 \quad (\text{Base Case}) \\
n \times (n-1)!, & \text{if } n > 1 \quad (\text{Recursive Step})
\end{cases}
$$

```python
def factorial(n: int) -> int:
    """
    Calculates the factorial of a non-negative integer n recursively.
    """
    if not isinstance(n, int) or n < 0:
        raise ValueError("Factorial is only defined for non-negative integers (n >= 0).")
    
    # Base case
    if n == 0 or n == 1:
        return 1
    
    # Recursive step
    return n * factorial(n - 1)

# Testing factorial
print(f"factorial(0) -> {factorial(0)}")
print(f"factorial(1) -> {factorial(1)}")
print(f"factorial(5) -> {factorial(5)} (5 * 4 * 3 * 2 * 1)")
print(f"factorial(7) -> {factorial(7)}")
```

## Question 9: Recursive Sum ($1 + 2 + 3 + \dots + n$)

### Objective:
Write a recursive function to calculate $1 + 2 + 3 + \dots + n$.

### Mathematical Definition:
$$
S(n) = \begin{cases}
0, & \text{if } n = 0 \\
1, & \text{if } n = 1 \quad (\text{Base Case}) \\
n + S(n-1), & \text{if } n > 1 \quad (\text{Recursive Step})
\end{cases}
$$

```python
def recursive_sum(n: int) -> int:
    """
    Calculates the sum of first n natural numbers recursively.
    """
    if not isinstance(n, int) or n < 0:
        raise ValueError("Input must be a non-negative integer.")
    
    # Base case
    if n == 0:
        return 0
    if n == 1:
        return 1
    
    # Recursive step
    return n + recursive_sum(n - 1)

# Testing recursive_sum
print(f"recursive_sum(1)  -> {recursive_sum(1)}")
print(f"recursive_sum(5)  -> {recursive_sum(5)} (1 + 2 + 3 + 4 + 5)")
print(f"recursive_sum(10) -> {recursive_sum(10)}")
print(f"recursive_sum(100)-> {recursive_sum(100)}")
```

## Question 10: Recursive Fibonacci

### Objective:
Write a recursive function to generate the Fibonacci sequence or calculate the $n$-th Fibonacci number.

### Mathematical Definition:
$$
F(n) = \begin{cases}
0, & \text{if } n = 0 \\
1, & \text{if } n = 1 \\
F(n-1) + F(n-2), & \text{if } n \ge 2
\end{cases}
$$

```python
def fibonacci(n: int) -> int:
    """
    Calculates the n-th Fibonacci number (0-indexed) using recursion.
    """
    if not isinstance(n, int) or n < 0:
        raise ValueError("Index n must be a non-negative integer.")
    
    # Base cases
    if n == 0:
        return 0
    if n == 1:
        return 1
    
    # Recursive step
    return fibonacci(n - 1) + fibonacci(n - 2)

def generate_fibonacci_sequence(count: int) -> list[int]:
    """
    Generates a list of the first 'count' Fibonacci numbers.
    """
    if count <= 0:
        return []
    return [fibonacci(i) for i in range(count)]

# Testing fibonacci & sequence generation
print(f"fibonacci(0) -> {fibonacci(0)}")
print(f"fibonacci(1) -> {fibonacci(1)}")
print(f"fibonacci(7) -> {fibonacci(7)}")
print(f"First 10 Fibonacci numbers: {generate_fibonacci_sequence(10)}")
```

## Question 11: Local vs Global Variable Scope Demonstration

### Objective:
Demonstrate local and global variable scope using a small program.

### Concept:
- **Global Scope:** Variables defined outside functions, accessible across modules and functions.
- **Local Scope:** Variables created within a function body, accessible only inside that function.
- **`global` keyword:** Allows modifying a global variable from within a local scope.

```python
# Global scope variables
app_version = "1.0.0"
counter = 100

def demonstrate_scope() -> None:
    """
    Demonstrates local scope, reading global variables, and modifying globals with 'global'.
    """
    # Local variable in demonstrate_scope
    local_message = "I am a local variable inside demonstrate_scope()"
    print(f"[Scope Demo] Local Variable: '{local_message}'")
    print(f"[Scope Demo] Reading Global Variable 'app_version': '{app_version}'")
    
    def modify_globals():
        global counter  # Modifying global variable
        counter += 25
        local_counter = 5  # Local to modify_globals
        print(f"  [Inner Func] Modified Global 'counter' to: {counter}")
        print(f"  [Inner Func] Inner Local 'local_counter': {local_counter}")

    modify_globals()
    print(f"[Scope Demo] Global Variable 'counter' after inner modification: {counter}")

# Demonstrating scope behavior
print("--- Before demonstrate_scope() ---")
print(f"Global counter: {counter}\n")
demonstrate_scope()
print("\n--- After demonstrate_scope() ---")
print(f"Global counter: {counter}")
```

## Question 12: Modular Mini Calculator

### Objective:
Create a mini calculator where each mathematical operation is implemented as a separate function and a main function controls the program.

### Architecture:
1. **Pure Operation Functions:** `calc_add`, `calc_subtract`, `calc_multiply`, `calc_divide`, `calc_power`, `calc_modulus`.
2. **Operation Registry:** Mapping menu options to functions and labels.
3. **Controller Function:** Orchestrates inputs, executes selected operation, and handles error states.

```python
# Operation functions
def calc_add(x: float, y: float) -> float:
    return x + y

def calc_subtract(x: float, y: float) -> float:
    return x - y

def calc_multiply(x: float, y: float) -> float:
    return x * y

def calc_divide(x: float, y: float) -> float | str:
    if y == 0:
        return "Error: Cannot divide by zero"
    return x / y

def calc_power(x: float, y: float) -> float:
    return x ** y

def calc_modulus(x: float, y: float) -> float | str:
    if y == 0:
        return "Error: Cannot perform modulus by zero"
    return x % y

# Operations Registry
OPERATIONS = {
    "1": ("Addition (+)", calc_add),
    "2": ("Subtraction (-)", calc_subtract),
    "3": ("Multiplication (*)", calc_multiply),
    "4": ("Division (/)", calc_divide),
    "5": ("Exponentiation (^)", calc_power),
    "6": ("Modulus (%)", calc_modulus),
}

def run_calculator_demo() -> None:
    """Automated demonstration of the mini calculator."""
    print("\n" + "=" * 50)
    print("         MINI CALCULATOR DEMO")
    print("=" * 50)
    test_cases = [
        ("1", 25, 15),
        ("2", 100, 42),
        ("3", 12, 8),
        ("4", 144, 12),
        ("4", 50, 0),     # Zero division test
        ("5", 2, 8),
        ("6", 29, 5),
    ]
    for choice, num1, num2 in test_cases:
        op_name, func = OPERATIONS[choice]
        result = func(num1, num2)
        print(f"  Operation: {op_name:<20} | Inputs: ({num1}, {num2}) -> Result: {result}")
    print("=" * 50)

# Run calculator demonstration
run_calculator_demo()
```
